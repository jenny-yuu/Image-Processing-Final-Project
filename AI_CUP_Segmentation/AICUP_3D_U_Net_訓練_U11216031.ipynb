```python
import gdown, os

os.makedirs('/content/CardiacSegV2/exps/exps/unet3d/chgh/tune_results', exist_ok=True)

url = "https://drive.google.com/uc?id=1aF8WoCgLeNMeuGF3fE4wc3lB0RMJU0ed"

out_path = "/content/CardiacSegV2/exps/exps/unet3d/chgh/tune_results/AICUP_training.zip"

gdown.download(url=url, output=out_path, quiet=False, fuzzy=True)
```

```python
import os
import json
import random
from glob import glob

# ================= 設定區 =================
base_dir = "/content/CardiacSegV2/dataset/chgh"
image_dir = os.path.join(base_dir, "training_image")
label_dir = os.path.join(base_dir, "training_label")
output_dir = "/content/CardiacSegV2/exps/data_dicts/chgh"
os.makedirs(output_dir, exist_ok=True)
# =========================================

# 1. 取得所有 ID (0001 ~ 0050)
# 這裡假設檔案都在 training_image 內，若您之前移動過檔案，建議先移回來
all_ids = []
for i in range(1, 51):
    pid = f"patient{str(i).zfill(4)}"
    # 檢查檔案是否存在 (防呆)
    img_path = os.path.join(image_dir, f"{pid}.nii.gz")
    if os.path.exists(img_path):
        all_ids.append(pid)
    else:
        # 如果檔案在 val 資料夾，嘗試去 val 抓 (兼容您之前的操作)
        val_img_path = os.path.join(base_dir, "val", f"{pid}.nii.gz")
        if os.path.exists(val_img_path):
             all_ids.append(pid)
             # 注意：如果是 K-Fold，建議邏輯上路徑要一致，
             # 但為了不報錯，這裡僅紀錄 ID，JSON路徑生成時需小心。

print(f"總共找到 {len(all_ids)} 位病人數據。")

# 2. 隨機打亂並固定種子 (確保之後可以重現)
random.seed(42)
random.shuffle(all_ids)

# 3. 分成 5 等份 (K=5)
num_folds = 5
fold_size = len(all_ids) // num_folds

for fold in range(num_folds):
    # 切分驗證集與訓練集
    val_start = fold * fold_size
    val_end = (fold + 1) * fold_size

    val_patients = all_ids[val_start:val_end]
    train_patients = [pid for pid in all_ids if pid not in val_patients]

    data_dict = {"train": [], "val": [], "test": []}

    # 輔助函式：回傳正確的檔案路徑
    def get_paths(pid):
        # 優先找 training_image
        img_p = os.path.join(image_dir, f"{pid}.nii.gz")
        lbl_p = os.path.join(label_dir, f"{pid}_gt.nii.gz")

        # 如果不在 training_image，就找 val (兼容您之前的操作)
        if not os.path.exists(img_p):
            img_p = os.path.join(base_dir, "val", f"{pid}.nii.gz")
            lbl_p = os.path.join(base_dir, "val", f"{pid}_gt.nii.gz")
        return img_p, lbl_p

    # 填入 Train
    for pid in train_patients:
        img, lbl = get_paths(pid)
        data_dict["train"].append({"image": img, "label": lbl})

    # 填入 Val
    for pid in val_patients:
        img, lbl = get_paths(pid)
        data_dict["val"].append({"image": img, "label": lbl})

    # 存檔
    json_filename = f"AICUP_fold{fold}.json"
    save_path = os.path.join(output_dir, json_filename)

    with open(save_path, "w") as f:
        json.dump(data_dict, f, indent=2)

    print(f"[{json_filename}] 生成完畢: Train: {len(train_patients)}, Val: {len(val_patients)}")

print("\n全部 JSON 生成完成！建議您訓練 5 次，每次換一個 JSON 檔。")
```

```python
import os
import zipfile

# ================= 參數設定區 =================
target_exp_name = 'AICUP_Fold0_Finetune' # 新的實驗名稱
target_json_name = 'AICUP_fold0.json'    # 新的 JSON
# ============================================

workspace_dir = '/content/CardiacSegV2'
model_name = 'unet3d'
data_name = 'chgh'

# 設定搜尋根目錄
root_exp_dir = os.path.join(workspace_dir, 'exps', 'exps', model_name, data_name, 'tune_results')
data_dicts_json = os.path.join(workspace_dir, 'exps', 'data_dicts', data_name, target_json_name)

# 確保目錄存在
os.makedirs(root_exp_dir, exist_ok=True)

# --- 步驟 1: 確保解壓縮 (如果找不到任何檔案才解壓) ---
zip_path = "/content/CardiacSegV2/exps/exps/unet3d/chgh/tune_results/AICUP_training.zip"

# 簡單檢查目錄是否幾乎是空的，如果是空的就強制解壓
if len(os.listdir(root_exp_dir)) <= 1 and os.path.exists(zip_path):
    print(f"目錄看起來是空的，正在解壓縮: {zip_path} ...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(root_exp_dir)
    print("解壓縮完成。")
else:
    print("目錄內已有檔案，跳過解壓縮。")

# --- 步驟 2: 暴力搜尋 best_model.pth ---
def find_best_model(search_dir):
    print(f"正在 {search_dir} 底下搜尋 best_model.pth ...")
    candidates = []

    # os.walk 會遍歷所有子資料夾
    for root, dirs, files in os.walk(search_dir):
        if 'best_model.pth' in files:
            full_path = os.path.join(root, 'best_model.pth')
            # 排除掉新實驗的資料夾 (避免抓到現在正在跑的空殼)
            if 'AICUP_training' in root:
                candidates.append(full_path)

    if not candidates:
        return None

    # 如果找到多個，通常選路徑最長的那個 (比較可能是深層的實驗目錄)
    # 或者單純選第一個，印出來給使用者看
    print(f"找到 {len(candidates)} 個權重檔：")
    for c in candidates:
        print(f" - {c}")

    return candidates[0] # 回傳第一個找到的

best_checkpoint = find_best_model(root_exp_dir)

if best_checkpoint:
    print(f"✅ 成功鎖定舊權重路徑: {best_checkpoint}")
else:
    print(f"❌ 嚴重錯誤：在 {root_exp_dir} 底下找不到任何 best_model.pth")
    # 嘗試列出目錄結構幫助除錯
    print("當前目錄結構如下：")
    !ls -R {root_exp_dir}

# --- 步驟 3: 設定變數供下一個訓練 Cell 使用 ---
exp_name = target_exp_name
data_dict_file_name = target_json_name
# data_dir 保持不變，指向 train 資料夾

print(f"-"*30)
print(f"🚀 準備開始訓練:")
print(f"   使用 JSON: {data_dicts_json}")
print(f"   新實驗名稱: {exp_name}")
print(f"   權重路徑: {best_checkpoint}")
```

```python
import os
import sys

# ================= 記憶體救援設定 =================
# 1. 告訴 Ray 不要太敏感 (原本 0.95 就殺，改成 0.99)
os.environ["RAY_memory_usage_threshold"] = "0.99"
# 2. 強制單執行緒 (避免多工吃光 RAM)
os.environ["OMP_NUM_THREADS"] = "1"
# =================================================

# 基礎路徑設定
workspace_dir = '/content/CardiacSegV2'
model_name = 'unet3d'
data_name = 'chgh'
base_exp_name = 'AICUP_Fold'
json_dir = os.path.join(workspace_dir, 'exps', 'data_dicts', data_name)

# 訓練參數 (已調整為省記憶體模式)
train_params = {
    "max_epoch": 100,
    "lr": 1e-4,
    "early_stop": 15,
    "roi_x": 128,
    "roi_y": 128,
    # Modify: 將 Z 軸從 120 大幅降低至 64，這是避免 OOM 的關鍵
    "roi_z": 64,
    "space_x": 0.7, "space_y": 0.7, "space_z": 1.0,
}

# 尋找舊權重 (如果重置環境後，您可能需要重新解壓縮 AICUP_training.zip)
# 這裡加入自動檢查與解壓功能
root_search_dir = os.path.join(workspace_dir, 'exps', 'exps', model_name, data_name, 'tune_results')
zip_path = "/content/CardiacSegV2/exps/exps/unet3d/chgh/tune_results/AICUP_training.zip"

def find_best_model(search_dir):
    for root, dirs, files in os.walk(search_dir):
        if 'best_model.pth' in files and 'AICUP_training' in root:
             return os.path.join(root, 'best_model.pth')
    return None

ssl_checkpoint = find_best_model(root_search_dir)

# 如果找不到權重，嘗試自動解壓 (因為您剛重置過環境)
if not ssl_checkpoint:
    print("⚠️ 找不到舊權重，嘗試解壓縮 AICUP_training.zip ...")
    import zipfile
    if os.path.exists(zip_path):
        with zipfile.ZipFile(zip_path, 'r') as zip_ref:
            zip_ref.extractall(root_search_dir)
        print("✅ 解壓縮完成，重新搜尋權重...")
        ssl_checkpoint = find_best_model(root_search_dir)
    else:
        print(f"❌ 錯誤：找不到 {zip_path}，請確認您有下載並上傳權重檔！")

if not ssl_checkpoint:
    raise FileNotFoundError("無法繼續：找不到 best_model.pth")

print(f"✅ 鎖定權重: {ssl_checkpoint}")

# 開始迴圈：從 Fold 0 跑到 Fold 4
# 如果您是跑到一半斷掉，可以把 range(5) 改成 range(斷掉的編號, 5)
for fold_idx in range(5):

    current_exp_name = f"{base_exp_name}{fold_idx}_Finetune"
    current_json = os.path.join(json_dir, f"AICUP_fold{fold_idx}.json")

    print(f"\n{'='*40}")
    print(f"🔥 啟動 Fold {fold_idx} (省記憶體模式)")
    print(f"   設定檔: {current_json}")
    print(f"{'='*40}\n")

    # 注意：移除了 --pin_memory
    command = (
        f"python /content/CardiacSegV2/expers/tune.py "
        f"--tune_mode=train "
        f"--exp_name={current_exp_name} "
        f"--data_name={data_name} "
        f"--data_dir=/content/CardiacSegV2/dataset/chgh/train "
        f"--root_exp_dir={root_search_dir} "
        f"--model_name={model_name} "
        f"--model_dir=./models "
        f"--log_dir=./logs "
        f"--eval_dir=./evals "
        f"--start_epoch=0 "
        f"--val_every=5 "
        f"--max_early_stop_count={train_params['early_stop']} "
        f"--max_epoch={train_params['max_epoch']} "
        f"--data_dicts_json={current_json} "
        # f"--pin_memory "  <-- 已移除此行以節省 RAM
        f"--out_channels=4 "
        f"--patch_size=4 "
        f"--feature_size=48 "
        f"--drop_rate=0.1 "
        f"--depths 3 3 9 3 "
        f"--kernel_size 7 "
        f"--exp_rate 4 "
        f"--norm_name='layer' "
        f"--a_min=-42 --a_max=423 "
        f"--space_x={train_params['space_x']} "
        f"--space_y={train_params['space_y']} "
        f"--space_z={train_params['space_z']} "
        f"--roi_x={train_params['roi_x']} "
        f"--roi_y={train_params['roi_y']} "
        f"--roi_z={train_params['roi_z']} "
        f"--optim='AdamW' "
        f"--lr={train_params['lr']} "
        f"--weight_decay=1e-5 "
        f"--ssl_checkpoint={ssl_checkpoint} "
        f"--use_init_weights "
        f"--infer_post_process "
    )

    exit_code = os.system(command)

    if exit_code != 0:
        print(f"⚠️ Fold {fold_idx} 訓練失敗 (Exit Code: {exit_code})")
    else:
        print(f"✅ Fold {fold_idx} 訓練完成！")
```
